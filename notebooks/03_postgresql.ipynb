# 03 — Esportazione PostgreSQL
Importiamo gli aggregati equivalenti, le 30 misure e il report qualità con lo stesso `run_id`. Le credenziali arrivano dalle variabili d'ambiente. La riesecuzione per lo stesso identificativo sostituisce le righe di quel run in una transazione. La query riepilogativa controlla gruppi e corse per categoria e metodo prima del commit.

```python
from IPython.display import Markdown, display

from src.postgresql import export_run, read_artifacts

quality, aggregates, timings = read_artifacts()
summary = export_run(quality, aggregates, timings)
print(f"Run esportato: {quality['run_id']}")

lines = ["| Categoria | Metodo | Gruppi | Corse |", "|---|---|---:|---:|"]
lines.extend(f"| {category} | {algorithm} | {groups} | {trips:,} |"
             for category, algorithm, groups, trips in summary)
display(Markdown("\n".join(lines)))
```
